**Code to check spelling in .txt files and find named entities**

The code below allows you to correct a Dutch text file and extract selected named entities (person names and place names). Execute the grey cells one after the other by pressing the <play> button. Wait for the green checkmark to appear before moving on to the next cell. You will be asked to upload a .txt file from your local machine in the process.

The code is written for use in Google Colab, so go to [colab.research.google.com](https://colab.research.google.com) and upload the code as a new Jupyter notebook file.

```python
# Install required libraries
!pip install -q pyspellchecker stanza

# Import required modules
import stanza
from spellchecker import SpellChecker
from google.colab import files
```

```python
# Set up Dutch spellchecker
spell = SpellChecker(language='nl')  # Dutch
print("Dutch spellchecker loaded with built-in dictionary.")
```

The next code section will perform the actual spell-checking, and depending on the size of your document, this can take up to an hour as the Google Colab environment is not very fast. Executing the code in a different environment can significantly speed up the spell check.

```python
# Function to correct OCR errors in Dutch text
def correct_ocr_errors(text):
    words = text.split()
    corrected_text = " ".join([spell.correction(word) if spell.correction(word) else word for word in words])
    return corrected_text

# Upload text file from user’s local machine
print("Please upload a .txt file with OCR text:")
uploaded = files.upload()

# Read uploaded file
filename = next(iter(uploaded))
with open(filename, "r", encoding="utf-8") as f:
    input_text = f.read()

# Correct OCR errors
corrected_text = correct_ocr_errors(input_text)
print("Corrected text preview:\n", corrected_text[:500])
```

```python
# Write corrected text to a new file

file_out="corrected_text.txt"
with open(file_out, "w", encoding="utf-8") as f:
    f.write(corrected_text)

print(f"Corrected text written to {file_out}")

# Download file
files.download(file_out)
```

The last section of code performs so-called Named Entity Recognition. This means that the code tries to identify words that fall into certain categories defined below. In the sample code here, the entities are "person" and "location". The script will output the entities found. As you will see, misidentifications are common, especially when foreign language words are in the input text.

```python
# Download Stanza Dutch model and load NLP pipeline
stanza.download("nl", verbose=False)
nlp = stanza.Pipeline(lang="nl", processors="tokenize,mwt,ner", verbose=False)

# Named Entity Recognition
def extract_named_entities(text):
    doc = nlp(text)
    named_entities = {"Person": [], "Location": []}
    for ent in doc.ents:
        if ent.type == "PER":
            named_entities["Person"].append(ent.text)
        elif ent.type in ["LOC", "GPE"]:
            named_entities["Location"].append(ent.text)
    return named_entities

# Apply NER to corrected text
entities = extract_named_entities(corrected_text)
for e in entities:
    print(f"{e}: {entities[e]}")
```
